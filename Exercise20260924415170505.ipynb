2. 為什麼自然數與整數都使用 `int` 表示？邏輯包容性： 自然數本質上就是「非負整數」（$\mathbb{N} \subset \mathbb{Z}$）。在計算機科學中，正整數與負整數在二進位運算、底層記憶體佈局上可以採用統一的規則（如二補數表示法）。介面簡化： 如果拆分為 Natural 與 Integer 兩個型別，當一個正整數進行減法結果變成負數時（如 $3 - 5 = -2$），程式就必須進行隱式的型別轉換。使用單一的 int 可以避免不必要的型別轉換開銷與複雜度。動態長度支援： Python 3 的 int 具備任意精度（Arbitrary-precision），能自動擴充記憶體以存放任意大小的整數，無論正負均可完美涵蓋。
3. Python 為什麼提供 `complex` 型別？科學計算與工程需求： 複數在訊號處理（如傅立葉轉換）、量子電腦、電路分析、流體力學及純數學運算中皆為核心工具。語法級原生支援： Python 設計初衷之一是作為科學家與工程師易於使用的語言，因此直接提供 complex 型別與虛數單位語法（如 1 + 2j），讓開發者無需額外撰寫類別（Class）即可進行直覺的向量與複數運算。
4. Python 為什麼沒有內建的有理數型別？效能與記憶體開銷： 嚴格的有理數需要同時維護分子與分母（並隨時進行最大公因數約分）。在連續計算中，分子分母的位數會呈指數級成長，導致記憶體消耗與計算速度急劇下降。float 已能滿足絕大多數需求： 現代硬體（CPU）皆內建浮點數運算單元（FPU），處理 IEEE 754 標準的 float 速度極快。雖然存在精準度誤差（如 $0.1 + 0.2 \neq 0.3$），但對於絕大多數應用情境已相當足夠。按需載入（Standard Library）： Python 遵循「內建核心保持精簡」的原則，將精準的分數運算放在標準庫 fractions 模組中（如 from fractions import Fraction），讓需要高精度分數運算的開發者自行匯入即可。

```python
format(0.1, ".20f")
```

1. 為什麼結果與數學上的答案不同？因為電腦底層採用 IEEE 754 標準的二進位浮點數（Binary Floating-point） 來儲存小數。在十進位中，許多看似簡短的有限小數（如 $0.1$ 與 $0.2$），轉換成二進位後會變成無限循環小數
2. 浮點數是否能精確表示所有小數？否。 浮點數只能精確表示分母為 2 的次方根（如 $1/2, 1/4, 1/8, 3/16$ 等）的十進位小數。對於絕大多數其他十進位小數（如 $0.1, 0.2, 0.3, 0.7$），在轉換為二進位時都會變成無窮循環，因而無法被 64 位元的浮點數結構「精確」儲存，只能儲存最接近的近似值。
3. 什麼是浮點數的精度（precision）？浮點數的精度是指該資料型別能夠精確記錄有效數字（Significant Digits）的位數極限。
4. 什麼是浮點數的範圍（range）？浮點數的範圍是指該型別能夠表示的「最大絕對值」與「最小非零絕對值」之間的數值區間（即能涵蓋多大或多小的數字）。
5. 精度與範圍有何不同？精度（Precision）決定數字能記得多「細」，範圍（Range）決定數字能算得多「大」。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？指的是「數值的位數長度不受到硬體暫存器（如 32-bit 或 64-bit）的固定限制」。
2. 它是否代表可以儲存無限大的整數？理論上可以，但實務上不行。
3. CPython 如何實現這項特性？動態位數陣列（ob_digit）：CPython 不把整數當作單一硬體數字，而是將其拆解成多個「小區塊（Digit）」存放在 C 語言的動態陣列中。在 64 位元系統上，每個 digit 預設佔用 30 個位元（即以 $2^{30}$ 為進位基底）。長度標頭（ob_size）：結構體內的 ob_size 記錄當前陣列用了幾個 digit 以及正負號。自動擴充：當進行加法或乘法導致數字超過當前容量時，CPython 會調用 PyMem_Realloc 動態向作業系統申請額外的記憶體，將陣列長度加長，並依序處理進位（Carry over）。

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？動態陣列結構： Python 的 int 在底層（CPython PyLongObject）採用動態陣列來記錄數字。
按需擴充記憶體： 它不使用硬體 CPU 固定的 32/64 位元暫存器，而是將數字切割成多個區塊（Digit）存放在記憶體中。當數值變大時，系統會自動分配更多記憶體來增長陣列長度，因此理論上沒有固定上限（僅受限於實體 RAM 大小）。
2. 為什麼 `float` 卻存在範圍限制？固定的硬體標準（IEEE 754）： float 嚴格遵循 IEEE 754 雙精度標準（Double Precision），每個 float 在記憶體中都被固定死為 64 個位元（8 Bytes）。位元分配受限： 這 64 個位元中有 11 個位元被固定分配給「指數部（Exponent）」。因為指數部的位元數有限，能代表的最大數量級被固定在 $10^{308}$ 左右，一旦超過就會觸發溢位（OverflowError 或變成 inf）。
3. 為什麼 `float` 可能產生誤差？二進位無法精確表示十進位小數： 電腦底層使用二進位（$2$ 的次方）儲存資料。許多十進位的有限小數（如 $0.1$ 或 $0.2$），在二進位中會變成無窮循環小數。有限尾數位的捨入截斷： 64 位元的 float 只分配了 52 個位元給「尾數部（Mantissa）」，無法完整存下無限循環的二進位位數，因此必須進行四捨五入（捨入截斷），這就產生了微小的浮點數誤差（如 0.1 + 0.2 == 0.3 判定為 False）。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。Arbitrary Precision（正確性優先）：核心哲學： 數據精準度絕不退讓。設計做法： 記憶體空間不設限，數值需要多少位數，系統就動態分配多少記憶體。運算速度可以慢，但結果必須達到 $100\%$ 精確，絕不允許任何捨入誤差或溢位。Limited Precision（效率優先）：核心哲學： 追求極致的計算速度與空間利用率，允許「足夠好」的微小誤差。設計做法： 將數值寫死在固定的記憶體位元數中（如 16 位元）。為了換取龐大的平行運算效能與吞吐量，主動放棄部分位數精度。
2. 各自的優點與缺點。Arbitrary Precision（任意精度）優點： 絕對精確無誤差、不會溢位爆掉。缺點： 極耗記憶體、運算速度慢（無法用 GPU 加速）。Limited Precision（有限精度）優點： 極省記憶體、速度極快（專為 GPU 平行運算設計）。缺點： 有微小捨入誤差、範圍過大過小容易溢位。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？顯存（VRAM）與頻寬瓶頸： 動輒數百億到上兆參數的大模型（如 GPT-4、Llama 3），若用 FP32 載入需數百 GB 甚至 TB 級顯存。改用 FP16/BF16 能直接將顯存需求砍半，讓模型能放進有限的 GPU 陣列中。
深度學習具備「抗噪韌性（Fault Tolerance）」： AI 模型的預測是基於機率與統計（矩陣乘法與權重加權），並不需要像銀行帳務那樣精確到每個小數點後第 20 位。微小的數值誤差不會改變模型的語意理解與生成能力。
GPU 硬體加速效益顯著： 現代 GPU（如 NVIDIA H100/B200）的 Tensor Core 針對 16 位元（FP16/BF16）甚至是 8 位元（FP8）運算進行了硬體級優化，計算速度比 FP32 快上數倍。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？顯存（VRAM）直接暴增 4 倍：原本需要 8 張 H100 就能訓練的模型，會變成需要 32 張，設備採購與維運成本呈指數級飆升。
訓練與推理速度斷崖式下跌： GPU 中的 FP64 核心數量遠少於 FP32/FP16，且無法發揮 Tensor Core 的矩陣加速優勢，模型訓練時間可能從「幾週」變成「幾年」。
巨大的資源浪費與高昂電費： 在對模型輸出品質（如回答精準度）幾乎沒有顯著提升的前提下，消耗數倍的電力與算力資源，完全不符合商業與工程效益。
