def get_zodiac(year):
    # 依照題目要求的順序排列十二生肖
    # 2004年為猴年，作為基準點 (2004 % 12 的餘數對應索引 0)
    zodiacs = [
        "猴 (monkey)", "雞 (rooster)", "狗 (dog)", "豬 (pig)",
        "鼠 (rat)", "牛 (ox)", "虎 (tiger)", "兔 (rabbit)",
        "龍 (dragon)", "蛇 (snake)", "馬 (horse)", "羊 (sheep)"
    ]
    
    # 判斷輸入為民國或西元
    # 若小於 1000 則視為民國年份，轉換為西元年
    if year < 1000:
        ad_year = year + 1911
        era_desc = f"民國 {year} 年 (西元 {ad_year} 年)"
    else:
        ad_year = year
        era_desc = f"西元 {ad_year} 年"
        
    # 計算生肖索引（2004年是猴，餘數為 0）
    index = (ad_year - 2004) % 12
    return era_desc, zodiacs[index]

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。金融領域（Financial Applications）：

Decimal 是極佳的方法。金融業對貨幣計算要求絕對精準（不能因為浮點數截斷而丟失任何一分錢），decimal 模組使用基數為 10 的十進位浮點運算，可完全避免十進位小數轉換為二進位產生的誤差。

科學計算（Scientific Computing）：

Decimal 不是通用最佳解。科學計算通常需要極高的計算效能與極大/極小的數值範圍，且運算涉及大量的矩陣乘法與微分方程。

使用 Decimal 會嚴重犧牲運算速度，且無法善用 CPU/GPU 的硬體浮點加速（FPU）。在科學計算中，通常依然使用 IEEE 754 浮點數（如 NumPy 的 float64），並透過數值分析技巧（如容許誤差 math.isclose / np.isclose）來控制誤差。
